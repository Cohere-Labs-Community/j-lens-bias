# Part 1 — Baseline walkthrough (confirmation step)
Sections 1-4 of walkthrough.ipynb, `MODEL_NAME = "Qwen/Qwen3.5-4B"`. The goal
is to confirm the lens sets up and the slice-vis page renders inline. This
section is context, not the graded deliverable.

T4 has no native bf16 tensor cores: load bf16 as in the walkthrough, and fall
back to float16 automatically if that errors. `compile=False` on Kaggle.


```python
import jlens

jlens.configure_logging()

MODEL_NAME = "Qwen/Qwen3.5-4B"
LENS_REPO = "neuronpedia/jacobian-lens"
LENS_REVISION = "qwen-n1000"
LENS_FILE = "qwen3.5-4b/jlens/Salesforce-wikitext/Qwen3.5-4B_jacobian_lens_n1000.pt"
```

```python
import time
import torch
import transformers

t0 = time.time()
try:
    hf_model = transformers.AutoModelForCausalLM.from_pretrained(
        MODEL_NAME, dtype=torch.bfloat16
    ).to(DEV)
    print("loaded in bf16")
except Exception as e:
    print("bf16 failed (%s); falling back to float16" % e)
    hf_model = transformers.AutoModelForCausalLM.from_pretrained(
        MODEL_NAME, dtype=torch.float16
    ).to(DEV)
    print("loaded in fp16")
tokenizer = transformers.AutoTokenizer.from_pretrained(MODEL_NAME)
print("model+tokenizer: %.0f s" % (time.time() - t0))

model = jlens.from_hf(hf_model, tokenizer, compile=False)
model
```

```python
lens = jlens.JacobianLens.from_pretrained(
    LENS_REPO, filename=LENS_FILE, revision=LENS_REVISION
)
lens
```

```python
prompt = "Fact: The currency used in the country shaped like a boot is"
layers = [
    model.n_layers // 4,
    model.n_layers // 2,
    model.n_layers // 4 * 3,
    model.n_layers - 2,
]

jlens_logits, model_logits, _ = lens.apply(model, prompt, layers=layers, positions=[-2])
logit_lens, _, _ = lens.apply(model, prompt, layers=layers, positions=[-2],
                              use_jacobian=False)


def top5(logits):
    return [tokenizer.decode([t]) for t in logits.topk(5).indices]


for layer in layers:
    print(f"L{layer:>3} logit-lens: {top5(logit_lens[layer][0])}")
    print(f"L{layer:>3} J-lens:     {top5(jlens_logits[layer][0])}")
print(f"model:           {top5(model_logits[0])}")
```

# Part 2 — Full-coverage lens.apply()
`positions=None` (verified in `jlens/lens.py`): every position `0..T-1`.
`layers=None` defaults to **all** fitted source layers, no stride (the
walkthrough's `layer_stride=2` was a demo setting only). One forward pass per
prompt returns the whole layer x position grid.

One full grid is rendered with `compute_slice` (no stride) and embedded.


```python
FULL_LAYERS = lens.source_layers          # every fitted layer, no stride
SKIP_FIRST = 16                           # fitting.py: SKIP_FIRST_N_POSITIONS
print("source layers:", FULL_LAYERS[0], "..", FULL_LAYERS[-1],
      "(", len(FULL_LAYERS), "layers )")
print(f"positions 0..{SKIP_FIRST - 1} are readable but outside the lens "
      "fitting distribution")


def read_full_grid(text):
    """One forward pass -> {layer: [seq_len, vocab] lens logits} for every
    fitted layer and every position, plus model final-layer logits."""
    jl, ml, ids = lens.apply(model, text, layers=FULL_LAYERS, positions=None,
                             use_jacobian=True)
    return jl, ml, ids
```

```python
# Render ONE full grid (every layer, every position) as a self-contained page.
slice_full = compute_slice(model, lens, prompt1, layer_stride=1,
                           mask_display=True)
page_full, _, _ = build_page(slice_full, prompt1,
                             title="full grid: every layer, every position",
                             description=("31 fitted layers, no stride, "
                                          "positions=None (all positions)"),
                             alt_token=gloss, mode="embed")
open("/kaggle/working/full_grid_embed.html", "w").write(page_full)
print(f"wrote /kaggle/working/full_grid_embed.html ({len(page_full) / 1e6:.1f} MB)")
notebook_iframe(page_full)
```

# Part 3 — Run on data/experiments prompts
Loader for the repo's own experiment prompt sets (README-verified format:
`{"items": [...]}`). Used set: `probe-swap.json`, 90 two-hop factual prompts.

Per `(prompt, layer, position)` cell the primary records are exactly the
task spec's definition:
- `top1_token`: argmax of the **lens readout** at that layer/position (raw
  argmax over the full vocabulary, no display mask);
- `rank_in_model`: that token's **full-vocabulary rank in the model's actual
  final-layer output distribution at the same position** (0 = the model's own
  top-1), computed with `jlens.vis._ranks_of` on the model logits;
- `matches_model_top1`: lens argmax == model argmax (the Part 4 statistic).

For one demonstration prompt the page-style rank is also shown, to make the
verified definition difference concrete: displayed token = word-masked argmax
of the LENS logits; its superscript rank = its full-vocab rank **in the lens
logits of that same layer** (NOT in the model distribution).


```python
import json
from pathlib import Path

EXP = Path("data/experiments")
print("available sets:", sorted(p.name for p in EXP.glob("*.json")))

PROMPT_SET = EXP / "probe-swap.json"
items = json.loads(PROMPT_SET.read_text(encoding="utf-8"))["items"]
print(f"{len(items)} prompts from {PROMPT_SET}")
print(json.dumps(items[0], indent=1))
```

```python
# Demonstration of the two rank definitions on one prompt, side by side.
from jlens.vis import _meaningful_token_mask, _ranks_of

jl, ml, ids = read_full_grid(items[0]["prompt"])
seq_len = int(ids.shape[1])
wordlike = _meaningful_token_mask(tokenizer, hf_model.config.vocab_size, DEV)

lay = 24                                   # a mid-stack layer
lens_l = jl[lay].to(DEV)
model_l = ml.to(DEV)
shown = lens_l.masked_fill(~wordlike, float("-inf")).argmax(-1)      # page display
page_rank = _ranks_of(lens_l, shown.unsqueeze(-1))[:, 0]             # in LENS logits
model_rank = _ranks_of(model_l, shown.unsqueeze(-1))[:, 0]           # in MODEL dist
raw_top1 = lens_l.argmax(-1)
raw_rank_in_model = _ranks_of(model_l, raw_top1.unsqueeze(-1))[:, 0]

for pos in range(max(0, seq_len - 6), seq_len):
    print(f"pos {pos:>3}  page: {tokenizer.decode([int(shown[pos])]):>12}^"
          f"{int(page_rank[pos]):<4} (lens-vocab rank)"
          f"   same token in model dist: rank {int(model_rank[pos])}"
          f"   raw lens argmax: {tokenizer.decode([int(raw_top1[pos])])!r}"
          f" -> model rank {int(raw_rank_in_model[pos])}")
print("\n=> page rank is within the LENS readout; Part 3's rank_in_model is"
      " within the MODEL's final-layer distribution. Different quantities.")
```

```python
import time
import torch
import pandas as pd

# Time estimate before committing: 3 prompts, then extrapolate.
def one_prompt_cells(idx):
    it = items[idx]
    jl, ml, ids = read_full_grid(it["prompt"])
    seq_len = int(ids.shape[1])
    model_l = ml.to(DEV)
    model_top1 = model_l.argmax(-1)
    rows = []
    for layer, logits in jl.items():
        lg = logits.to(DEV)
        top1 = lg.argmax(-1)                                   # [seq_len]
        rank_in_model = _ranks_of(model_l, top1.unsqueeze(-1))[:, 0]
        for pos in range(seq_len):
            rows.append(dict(
                prompt_idx=idx, name=it.get("name", ""), layer=int(layer),
                pos=pos, pos_from_end=pos - seq_len,
                pos_ge16=pos >= SKIP_FIRST,
                top1_id=int(top1[pos]),
                top1_token=tokenizer.decode([int(top1[pos])],
                                            clean_up_tokenization_spaces=False),
                rank_in_model=int(rank_in_model[pos]),
                matches_model_top1=bool(top1[pos] == model_top1[pos]),
            ))
        del lg
    return rows

t0 = time.time()
for k in range(3):
    one_prompt_cells(k)
per = (time.time() - t0) / 3
print(f"~{per:.2f} s/prompt on this GPU -> {per * len(items) / 60:.1f} min "
      f"for all {len(items)} prompts (plus ~1 min model/lens load, already done)")
```

```python
# Full run. All 90 prompts at ~0.5-2 s/prompt (est. 1-3 min of GPU time on
# this class of accelerator, measured in the timing cell above). If the
# session is short, cap MAX_PROMPTS; coverage is reported below either way.
MAX_PROMPTS = len(items)

cells = []
t0 = time.time()
for idx in range(min(MAX_PROMPTS, len(items))):
    cells.extend(one_prompt_cells(idx))
    if (idx + 1) % 10 == 0:
        print(f"  {idx + 1}/{min(MAX_PROMPTS, len(items))} prompts  "
              f"{(time.time() - t0) / (idx + 1):.2f} s/prompt  {len(cells):,} cells")

df = pd.DataFrame(cells)
print(f"\n{len(df):,} cells = {df.prompt_idx.nunique()} prompts x "
      f"{df.layer.nunique()} layers x positions")
df.to_csv("/kaggle/working/probe_swap_lens_grid.csv", index=False)
print("wrote /kaggle/working/probe_swap_lens_grid.csv")
n_cov = df.prompt_idx.nunique()
coverage = "all" if n_cov == len(items) else f"capped by MAX_PROMPTS={MAX_PROMPTS}"
print(f"coverage: {n_cov} of {len(items)} experiment prompts ({coverage})")
```

# Part 4 — Per-layer summary table
Primary statistic per the task spec: per-layer **top-1 accuracy of the lens
against the model's real output** (`matches_model_top1`: lens argmax equals
model argmax), the same shape as the paper's layer-wise accuracy curve. Also
reported: the median rank (in the model distribution) of the lens's top-1,
and top-5 / top-10 hit rates via `rank_in_model`. Each is given over all
positions and over positions >= 16 (inside the lens fitting distribution)
separately, per the verified position caveat.


```python
import numpy as np

def layer_summary(g):
    r = g["rank_in_model"].to_numpy()
    r16 = g.loc[g["pos_ge16"], "rank_in_model"].to_numpy()
    return pd.Series({
        "cells": len(g),
        "top1_acc": 100.0 * g["matches_model_top1"].mean(),
        "top5_hit": 100.0 * (r < 5).mean(),
        "top10_hit": 100.0 * (r < 10).mean(),
        "med_rank_in_model": np.median(r),
        "top1_acc_pos16plus": (100.0 * g.loc[g["pos_ge16"],
                                             "matches_model_top1"].mean()
                               if r16.size else np.nan),
        "med_rank_pos16plus": np.median(r16) if r16.size else np.nan,
    })

summary = df.groupby("layer").apply(layer_summary, include_groups=False)
print(f"per-layer summary over {df.prompt_idx.nunique()} prompts, "
      f"{len(df):,} cells")
print(summary.round(2).to_string())   # printed so it lands in the kernel log
summary.round(2)
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
ax = axes[0]
ax.plot(summary.index, summary["top1_acc"], marker="o", ms=3,
        label="lens top-1 == model top-1")
ax.plot(summary.index, summary["top10_hit"], marker="s", ms=3,
        label="lens top-1 in model top-10")
ax.set_xlabel("layer"); ax.set_ylabel("% of cells")
ax.set_title("Lens vs model output, by layer")
ax.legend(fontsize=8); ax.grid(alpha=0.3)

ax = axes[1]
ax.plot(summary.index, summary["med_rank_in_model"] + 1, marker="o", ms=3,
        label="all positions")
ax.plot(summary.index, summary["med_rank_pos16plus"] + 1, marker="s", ms=3,
        label="positions >= 16 (lens fitting distribution)")
ax.set_yscale("log"); ax.set_xlabel("layer")
ax.set_ylabel("median rank in model dist + 1")
ax.set_title("Lens top-1 token's rank within the model output")
ax.legend(fontsize=8); ax.grid(alpha=0.3)
fig.tight_layout()
fig.savefig("/kaggle/working/layer_summary.png", dpi=150)
print("wrote /kaggle/working/layer_summary.png")
plt.show()
```

# Observations

Actually just last week i was working on implementing this repo for Gemma 4 E4b model and i had some interesting results so though of sharing them while i cant share the code as of its a part of some project so here are some interesting abservations from gemma 4 E4B model 

Both models consolidate the same way: near-chance readout through the early and mid stack,then a sharp break into committed, model-matching output. Qwen3.5-4B's jump concentratesaround L22–23 of 32 layers; Gemma-4 E4B's sat at L23 of 42.

Gemma-4 E4B (42 layers) and Qwen3.5-4B (32 layers) both have residual width 2560, and both
are hybrid stacks where only some layers do full global attention. Qwen3.5's only global
layers are 3, 7, 11, ..., 31 — layer 23 is one of them. Gemma-4 alternates local and global,
which also puts 23 on a global layer. So the readout in both models sharpens right where a
global layer writes long-range content into the residual stream. The shared mechanism is the
binding step, not the integer — at a different width or architecture the knee would land
elsewhere.
