```python
import torch
import transformers
import jlens
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig

model_id = "Qwen/Qwen3.6-27B" # Aligned with lens checkpoint version

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16
)

tok = AutoTokenizer.from_pretrained(model_id)
hf = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map="auto",
    trust_remote_code=True
)

model = jlens.from_hf(hf, tok)


```

```python
lens = jlens.JacobianLens.from_pretrained(
    "neuronpedia/jacobian-lens", 
    filename="qwen3.6-27b/jlens/Salesforce-wikitext/Qwen3.6-27B_jacobian_lens_n1000.pt",
    revision="qwen-n1000"
)
```

```python
prompt = "Fact: The currency used in the country shaped like a boot is"

#No need to track gradient 
with torch.no_grad():
    # Use -1 to inspect the prediction following "is" (e.g. " Euro")
    lens_logits, model_logits, _ = lens.apply(model, prompt, positions=[-2])

for layer, logits in sorted(lens_logits.items()):
    tokens = [tok.decode([t]) for t in logits[0].topk(5).indices]
    print(f"Layer {layer:02d}: {tokens}")
```

```python
import gzip
import json

from jlens.examples import EXAMPLES, resolve_prompt
from jlens.vis import build_page, compute_slice, notebook_iframe

# English gloss for Qwen's Chinese/Japanese/Korean vocab tokens (machine-
# generated, best-effort), shown next to
# the token in the page (alt_token=).
gloss = {
    int(k): v for k, v in json.load(gzip.open("assets/qwen_gloss.json.gz")).items()
}

example = next(e for e in EXAMPLES if e.slug == "multihop")
prompt = resolve_prompt(example, tok)

slice_data = compute_slice(
    model,
    lens,
    prompt,
    layer_stride=2,
    # Empirically on Qwen, the interesting word tokens trail punctuation and
    # single-character tokens in the raw top-K; mask to word-like tokens only.
    mask_display=True,
)
page, _, _ = build_page(
    slice_data,
    prompt,
    title=example.section,
    description=example.description,
    alt_token=gloss,
)
notebook_iframe(page)
```

```python
import json
from collections import Counter

with open("data/experiments/probe-swap.json") as file:
    raw = json.load(file)

probe_swap_data = raw["items"]

prompts = [entry["prompt"] for entry in probe_swap_data]

tokenizer = model.tokenizer

layer_counts: dict[int, Counter] = {}
position_layer_counts: dict[tuple[int, int], Counter] = {}

for prompt in prompts:
    input_ids = model.encode(prompt)
    seq_len = input_ids.shape[1]

    lens_logits, model_logits, _ = lens.apply(model, prompt, positions=list(range(seq_len)))

    for layer, logits in lens_logits.items():
        top1_per_position = logits.argmax(dim=-1).tolist()
        layer_counts.setdefault(layer, Counter()).update(top1_per_position)
        for position, token_id in enumerate(top1_per_position):
            position_layer_counts.setdefault((position, layer), Counter())[token_id] += 1
```

```python
import pandas as pd
def decode(token_id: int) -> str:
    return tokenizer.decode([token_id], clean_up_tokenization_spaces=False)


df_by_layer = (
    pd.DataFrame(
        {
            "layer": layer,
            "top_token": decode(counter.most_common(1)[0][0]),
            "count": counter.most_common(1)[0][1],
            "total_seen": sum(counter.values()),
        }
        for layer, counter in sorted(layer_counts.items())
    )
    .set_index("layer")
)


df_grid_long = pd.DataFrame(
    {"position": position, "layer": layer, "top_token": decode(counter.most_common(1)[0][0])}
    for (position, layer), counter in position_layer_counts.items()
)
df_by_position = df_grid_long.pivot(index="position", columns="layer", values="top_token").sort_index()
```

# **Experimenting with vector swapping:**

```python
#Preliminary algorithm for swapping J space vectors with adjustable strength based on the J-space paper.

W_U = hf.get_output_embeddings().weight #Get unembedding vector

def get_lens_vectors(layer, source_id, target_id):
    '''Get relevant J space vectors. The J lens vector is pre transposed at storage.'''
    J_l = lens.J[layer]
    v_s = W_U[source_id] @ J_l 
    v_t = W_U[target_id] @ J_l
    return v_s, v_t
    
def make_swap_hook(v_s, v_t, positions=None, alpha=1.0):
    '''The idea is like the one of the paper, the goal is to break up the hidden state h into the c1 * source vector + c2 * target vector + w
        where w is orthogonal. This is so we can perform magnitude aware swapping where we simply swap the proportions of each vector in h 
        to both source and target vectors a method of doing so involves le
    
    '''    
    V = torch.stack([v_s, v_t], dim=1)
    V_pinv = torch.linalg.pinv(V) #Get pseudo inverse of the source and target vectors,the idea is treating h like a linear regression problem and solving via least squares using pseeudo inverese
    def hook(module, input, output):
        h = output[0] if isinstance(output, tuple) else output # Shape (1,seq,d_model) batch == 1 for inference.
        pos = positions if positions is not None else slice(None)
        h_sel = h[:, pos, :] 
        c = h_sel @ V_pinv.T   #Getting coeffecients, how much of v_s is in selected h, how much of v_t is in selcted h Shape (batch(1), n_pos, 2)
        c_swapped = c.flip(dims=[-1]) * alpha + c * (1 - alpha)  # swap w/ optional strength
        delta = (c_swapped - c) @ V.T       # Actual scaled vectors to be added/subtracted after calculating c values

        h[:, pos, :] = h_sel + delta #

        if isinstance(output, tuple):
            return (h,) + output[1:]
        return h
        
    return hook
```
